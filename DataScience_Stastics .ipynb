# 2. Standard Deviation Illustration
x = np.linspace(-4, 4, 1000)
y = 1 / np.sqrt(2 * np.pi) * np.exp(-x**2 / 2)

plt.figure(figsize=(6, 4))
plt.plot(x, y)
plt.fill_between(x, y, where=(x >= -1) & (x <= 1), color='green', alpha=0.3, label='±1 SD (~68%)')
plt.fill_between(x, y, where=(x >= -2) & (x <= 2), color='orange', alpha=0.2, label='±2 SD (~95%)')
plt.fill_between(x, y, where=(x >= -3) & (x <= 3), color='red', alpha=0.1, label='±3 SD (~99.7%)')
plt.title('Standard Deviation in Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
std_dev_path = "/mnt/data/std_deviation.png"
plt.savefig(std_dev_path)
plt.close()

# 📊 Detailed Statistics Notes for Data Science

These notes aim to provide a strong foundation in statistics with intuitive analogies and detailed visual explanations.

---

## 1. **What is Statistics?**

**Definition:**  
Statistics is the science of collecting, analyzing, interpreting, presenting, and organizing data.

**Analogy:**  
Think of statistics like organizing your closet. You sort clothes (collect data), find trends (analyze), and decide what you need to buy more (make inferences).

---

## 2. **Types of Statistics**

- **Descriptive Statistics**: Deals with summarizing and organizing data.
- **Inferential Statistics**: Uses a sample of data to make inferences about a population.

**Analogy:**  
Descriptive statistics is like taking attendance in your class. Inferential statistics is like estimating how many students will pass the final exam based on a sample test.

**Figure: Descriptive vs Inferential**  
![Descriptive vs Inferential](https://upload.wikimedia.org/wikipedia/commons/thumb/e/e8/Descriptive_vs_inferential_statistics.jpg/640px-Descriptive_vs_inferential_statistics.jpg)

---

## 3. **Central Tendency**

- **Mean**: The average of all numbers.
- **Median**: The middle value when sorted.
- **Mode**: The most frequent number.

**Analogy:**  
If three friends get 80, 85, and 90 marks:  
- Mean = (80+85+90)/3 = 85  
- Median = 85 (middle)  
- Mode = most repeated score

---

## 4. **Dispersion Measures**

These show how much the data is spread out:

- **Range** = Max - Min
- **Variance** = Average squared deviation from the mean
- **Standard Deviation (SD)** = √Variance

**Figure: Standard Deviation Explained**  


**Explanation:**  
- Green region: ~68% data falls within 1 SD of the mean  
- Orange region: ~95% falls within 2 SD  
- Red region: ~99.7% falls within 3 SD  
This tells you how "tight" your data is around the mean.

---

## 5. **Probability**

**Definition:**  
Chance that an event will occur, ranging from 0 (impossible) to 1 (certain).

**Analogy:**  
Tossing a fair coin gives a 0.5 chance (50%) of heads.

---

## 6. **Normal Distribution**

**Definition:**  
A symmetric bell-shaped curve where most of the data lies near the mean.

**Figure: Normal Distribution Curve**  


**Explanation:**  
This curve shows that values near the mean are more frequent in occurrence than values far from the mean. Most natural phenomena follow this pattern (e.g., height, weight).

---

## 7. **Correlation vs Causation**

- **Correlation**: Two variables move together.
- **Causation**: One variable causes a change in another.

**Analogy:**  
Eating ice cream and drowning rates rise in summer. They correlate, but one doesn't cause the other.

---

## 8. **Hypothesis Testing**

**Definition:**  
A method to test assumptions (hypotheses) using statistical evidence.

- **Null Hypothesis (H0)**: No effect.
- **Alternative Hypothesis (H1)**: There is an effect.

**Analogy:**  
A person is innocent until proven guilty — the null hypothesis is that they're innocent.

---

## 9. **Confidence Interval**

**Definition:**  
A range of values derived from a sample that’s likely to include the population parameter.

**Analogy:**  
You can say with 95% confidence that your friend’s average marks lie between 75–85.

---

## 10. **P-Value**

**Definition:**  
The probability of getting the observed result assuming the null hypothesis is true.

**Analogy:**  
A low p-value (e.g., <0.05) means the result is unlikely due to chance — it's like saying, “there’s strong evidence against the null.”

---

**Happy Studying! 🧠**
