✅ Problem Recap:
Given a list nums and a target value target, return the indices of the two numbers in nums that add up to target.

💡 Intuition:
We want to find two numbers such that:

nums[i] + nums[j] == target
We can do this using a dictionary (hashmap) to store numbers we've seen so far and check if the complement exists.

```python
def twoSum(nums, target):
    num_map = {}  # to store number:index pairs
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]  # return indices
        num_map[num] = i  # store index of the current number
```

```python
nums = [2, 7, 11, 15]
target = 9

# First iteration:
# num = 2, complement = 7 → not in map → store 2:0

# Second iteration:
# num = 7, complement = 2 → found in map → return [0, 1]
```

```python
print(twoSum([2, 7, 11, 15], 9))  # Output: [0, 1]

```

Problem:
Given a string s containing just the characters '(', ')', '{', '}', '[', and ']', determine if the input string is valid.

An input string is valid if:

Open brackets must be closed by the same type of brackets.

Open brackets must be closed in the correct order.


Input: s = "()"            → Output: True  
Input: s = "()[]{}"        → Output: True  
Input: s = "(]"            → Output: False  
Input: s = "([)]"          → Output: False  
Input: s = "{[]}"          → Output: True


```python
def isValid(s):
    stack = []
    bracket_map = {')': '(', ']': '[', '}': '{'}

    for char in s:
        if char in bracket_map.values():
            # If it's an opening bracket, push to stack
            stack.append(char)
        elif char in bracket_map:
            # If it's a closing bracket, check for matching top of stack
            if not stack or stack[-1] != bracket_map[char]:
                return False
            stack.pop()  # matched pair, remove opening
        else:
            return False  # Invalid character

    # Valid only if no unmatched brackets remain
    return len(stack) == 0

```

```python
print(isValid("()"))        # True
print(isValid("()[]{}"))    # True
print(isValid("(]"))        # False
print(isValid("([)]"))      # False
print(isValid("{[]}"))      # True
```
